### Feature Descriptions

**Demographic:**
- `school`: Student's school (GP or MS)
- `sex`: Student's sex (F or M)
- `age`: Student's age (15 to 22)
- `address`: Home address type (U=urban, R=rural)

**Family:**
- `famsize`: Family size (LE3: ≤3 or GT3: >3)
- `Pstatus`: Parent's cohabitation status (T=together, A=apart)
- `Medu`: Mother's education (0-4, higher = more education)
- `Fedu`: Father's education (0-4, higher = more education)

**School-related:**
- `traveltime`: Home to school travel time (1=<15 min, 2=15-30 min, 3=30-60 min, 4=>60 min)
- `studytime`: Weekly study time (1=<2 hours, 2=2-5 hours, 3=5-10 hours, 4=>10 hours)
- `failures`: Number of past class failures (0-3)
- `schoolsup`: Extra educational support (yes or no)
- `famsup`: Family educational support (yes or no)
- `paid`: Extra paid classes (yes or no)
- `activities`: Extra-curricular activities (yes or no)
- `higher`: Wants to take higher education (yes or no)
- `internet`: Internet access at home (yes or no)

**Social:**
- `romantic`: In a romantic relationship (yes or no)
- `famrel`: Quality of family relationships (1-5, higher = better)
- `freetime`: Free time after school (1-5, higher = more)
- `goout`: Going out with friends (1-5, higher = more)
- `Dalc`: Workday alcohol consumption (1-5, higher = more)
- `Walc`: Weekend alcohol consumption (1-5, higher = more)
- `health`: Current health status (1-5, higher = better)

**Academic Performance:**
- `absences`: Number of school absences (0 to 93)
- `G1`: First period grade (0 to 20)
- `G2`: Second period grade (0 to 20)
- `G3`: **Final grade (0 to 20) - THIS IS OUR TARGET!**
